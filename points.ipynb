```python
import pandas as pd # work with data
import numpy as np # use calculatons
import matplotlib.pyplot as plt # use ploting graphs
import seaborn as sns
```

```python
df = pd.read_csv("/content/cleaned_insurance.csv")
```

```python
df["ChargesLog"] = np.log(df["charges"] + 1)
```

```python
plt.figure(figsize=(15, 10))

# Charges Distribution
plt.subplot(2, 3, 1)
plt.hist(df["charges"])
plt.title("Charges Distribution")
plt.xlabel("Charges")
plt.ylabel("Count")

# Age Distribution
plt.subplot(2, 3, 2)
plt.hist(df["age"])
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Count")

# BMI vs Charges
plt.subplot(2, 3, 3)
plt.scatter(df["bmi"], df["charges"])
plt.title("BMI vs Charges")
plt.xlabel("BMI")
plt.ylabel("Charges")

# Charges by Children
plt.subplot(2, 3, 4)
plt.boxplot([df[df["children"] == i]["charges"] for i in df["children"].unique()])
plt.title("Charges by Number of Children")
plt.xlabel("Children")
plt.ylabel("Charges")

# Log Charges
plt.subplot(2, 3, 5)
plt.hist(df["ChargesLog"])
plt.title("Log Charges")
plt.xlabel("Log Charges")
plt.ylabel("Count")

plt.tight_layout()
plt.show()
```

### exploring one topic deeper

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the cleaned insurance dataset
df = pd.read_csv("/content/cleaned_insurance.csv")
df.head()

# Check BMI effect across age groups
sns.scatterplot(
    x="bmi",
    y="charges",
    hue="smoker",
    size="age",
    sizes=(20,200),
    data=df,
    alpha=0.5
)
plt.title("BMI vs Charges (Color=Smoker, Size=Age)")
plt.show()

# Insight:
# Higher BMI generally increases charges
# Smokers with high BMI and older age have the highest charges
# Combination of multiple factors strongly affects charges
```
